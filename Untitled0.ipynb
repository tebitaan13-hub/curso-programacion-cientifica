Ejercicio 1. Ficha de la campaña de medición

```python
codigo_estacion = "EST-01"
sector = "Norte"
altitud_m = 1495
latitud = 6.28
sensor_calibrado = True
numero_lecturas = 305
factor_conversion = 1000

print("Código:", codigo_estacion, type(codigo_estacion))
print("Sector:", sector, type(sector))
print("Altitud:", altitud_m, type(altitud_m))
print("Latitud:", latitud, type(latitud))
print("Sensor calibrado:", sensor_calibrado, type(sensor_calibrado))
print("Número de lecturas:", numero_lecturas, type(numero_lecturas))
print("Factor de conversion:", factor_conversion, type(factor_conversion))
```

Ejercicio 2. Conversión de unidades y error

```python
lectura_sensor_mg = 0.0418
valor_patron_ug = 40.0

lectura_sensor_ug = lectura_sensor_mg * factor_conversion
error_absoluto = lectura_sensor_ug - valor_patron_ug
error_relativo_porcentual = (error_absoluto / valor_patron_ug) * 100
error_cuadratico = error_absoluto ** 2

print("Lectura convertida (ug/m3):", lectura_sensor_ug)
print("Error absoluto:", error_absoluto)
print("Error relativo porcentual:", error_relativo_porcentual, "%")
print("Error cuadrático:", error_cuadratico)
```

Ejercicio 3. Cobertura temporal del lote

```python
lecturas_por_hora = 12

horas_completas = numero_lecturas // lecturas_por_hora
lecturas_sobrantes = numero_lecturas % lecturas_por_hora

dias_completos = horas_completas // 24
horas_sobrantes = horas_completas % 24

print("Horas completas:", horas_completas)
print("Lecturas sobrantes:", lecturas_sobrantes)
print("Días completos:", dias_completos)
print("Horas sobrantes:", horas_sobrantes)
```

Ejercicio 4. Validar una lectura

```python
lectura = 41.8
codigo_dato_faltante = -999.0
limite_superior_sensor = 500.0

es_dato_faltante = lectura == codigo_dato_faltante
esta_en_rango = lectura >= 0 and lectura <= limite_superior_sensor
lectura_valida = esta_en_rango and not es_dato_faltante

print("¿Es dato faltante?:", es_dato_faltante)
print("¿Está en rango?:", esta_en_rango)
print("¿Lectura válida?:", lectura_valida)
```

Ejercicio 5. Clasificar la lectura en la escala de calidad del aire

```python
if not lectura_valida:
    categoria = "Lectura no válida"
elif lectura <= 12.0:
    categoria = "Buena"
elif lectura <= 35.4:
    categoria = "Moderada"
elif lectura <= 55.4:
    categoria = "Dañina para grupos sensibles"
else:
    categoria = "Dañina para la salud"

print("Categoría de calidad del aire:", categoria)
```

Ejercicio 6. Limpiar la serie horaria

```python
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6,
                 -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]

lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

for lectura_actual in lecturas_pm25:
    if lectura_actual == -999.0:
        lecturas_descartadas += 1
        continue

    lecturas_validas.append(lectura_actual)
    suma_pm25 += lectura_actual

promedio_pm25 = suma_pm25 / len(lecturas_validas)

print("Lecturas válidas:", len(lecturas_validas))
print("Lecturas descartadas:", lecturas_descartadas)
print("Promedio PM2.5:", promedio_pm25)
print("Serie limpia:", lecturas_validas)
```

Ejercicio 7. Estadística descriptiva de la serie limpia

```python
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for valor in lecturas_validas:
    if valor > maximo_pm25:
        maximo_pm25 = valor

    if valor < minimo_pm25:
        minimo_pm25 = valor

rango_pm25 = maximo_pm25 - minimo_pm25

suma_desviaciones = 0.0

for valor in lecturas_validas:
    suma_desviaciones += (valor - promedio_pm25) ** 2

desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

print("Máximo PM2.5:", maximo_pm25)
print("Mínimo PM2.5:", minimo_pm25)
print("Rango PM2.5:", rango_pm25)
print("Desviación estándar:", desviacion_estandar)
```

Ejercicio 8. Decaimiento tras cerrar la fuente de emisión

```python
concentracion = 58.3
factor_decaimiento = 0.85
umbral_buena = 12.0
horas_transcurridas = 0

while concentracion > umbral_buena:
    concentracion = concentracion * factor_decaimiento
    horas_transcurridas += 1

print("Horas transcurridas:", horas_transcurridas)
print("Concentración final:", concentracion)
```

Ejercicio 9. Registros de varias estaciones

```python
registros_estaciones = [
    {
        "codigo": "EST-01",
        "sector": "Norte",
        "coordenadas": (6.28, -75.57),
        "lecturas": [8.4, 12.7, 15.2]
    },
    {
        "codigo": "EST-02",
        "sector": "Centro",
        "coordenadas": (6.24, -75.58),
        "lecturas": [41.8, 58.3, 36.1]
    },
    {
        "codigo": "EST-03",
        "sector": "Sur",
        "coordenadas": (6.19, -75.59),
        "lecturas": [9.6, 13.9, 7.5],
        "humedad_relativa": 68.0
    }
]

for estacion in registros_estaciones:
    latitud, longitud = estacion["coordenadas"]

    suma_lecturas = 0.0

    for valor in estacion["lecturas"]:
        suma_lecturas += valor

    promedio_estacion = suma_lecturas / len(estacion["lecturas"])
    humedad = estacion.get("humedad_relativa", "no disponible")

    if promedio_estacion <= 12.0:
        categoria_estacion = "Buena"
    elif promedio_estacion <= 35.4:
        categoria_estacion = "Moderada"
    elif promedio_estacion <= 55.4:
        categoria_estacion = "Dañina para grupos sensibles"
    else:
        categoria_estacion = "Dañina para la salud"

    print(
        f"Código: {estacion['codigo']} | "
        f"Sector: {estacion['sector']} | "
        f"Latitud: {latitud} | "
        f"Longitud: {longitud} | "
        f"Promedio: {promedio_estacion:.2f} | "
        f"Categoría: {categoria_estacion} | "
        f"Humedad: {humedad}"
    )
```
