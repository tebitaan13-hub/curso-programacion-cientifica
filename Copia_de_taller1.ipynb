1 - En este ejercicio se declaran las variables de los productos y se verifica el tipo de dato de cada una

```python
nombre_producto = "Cuaderno cuadriculado"
precio_cuaderno = 3500
precio_lapiz = 800
precio_cafe = 6200.50
hay_descuento = False
cantidad_cuadernos = 3
cantidad_lapices = 5

print(type(nombre_producto))
print(type(precio_cuaderno))
print(type(precio_lapiz))
print(type(precio_cafe))
print(type(hay_descuento))
print(type(cantidad_cuadernos))
print(type(cantidad_lapices))
```

Ejercicio 2
Se calcula el costo de los cuadernos y lápices, el total de la compra incluyendo el café y el descuento del 10 %

```python
total_cuadernos = precio_cuaderno * cantidad_cuadernos
total_lapices = precio_lapiz * cantidad_lapices

total_general = total_cuadernos + total_lapices + precio_cafe
descuento = total_general * 0.10
total_con_descuento = total_general - descuento

print("Total de cuadernos:", total_cuadernos)
print("Total de lápices:", total_lapices)
print("Total general:", total_general)
print("Descuento del 10%:", descuento)
print("Total con descuento:", total_con_descuento)
```

Ejercicio 3
Se calcula cuántos paquetes completos de cuatro cuadernos se pueden formar, cuántos sobran y el precio del café elevado al cuadrado

```python
paquetes_completos = cantidad_cuadernos // 4
cuadernos_sueltos = cantidad_cuadernos % 4
cafe_al_cuadrado = precio_cafe ** 2

print("Paquetes completos:", paquetes_completos)
print("Cuadernos sueltos:", cuadernos_sueltos)
print("Precio del café al cuadrado:", cafe_al_cuadrado)
```

Ejercicio 5
Se calcula el promedio de las temperaturas de tres días y se realizan comparaciones con la temperatura de referencia de 20 grados

```python
temp_lunes = 18.5
temp_martes = 21.0
temp_miercoles = 19.8
temp_promedio_semana = 20.0

promedio_temperaturas = (
    temp_lunes + temp_martes + temp_miercoles
) / 3

diferencia_temperaturas = temp_martes - temp_lunes

print("Promedio de los tres días:", promedio_temperaturas)
print("Diferencia entre martes y lunes:", diferencia_temperaturas)
print("¿Martes superó el promedio de referencia?",
      temp_martes > temp_promedio_semana)
print("¿Miércoles fue igual al promedio de referencia?",
      temp_miercoles == temp_promedio_semana)
print("¿Lunes fue el día más frío?",
      temp_lunes < temp_martes and temp_lunes < temp_miercoles)
```

Ejercicio 7
Se calcula el costo del consumo eléctrico y se comprueba si está dentro del rango normal de 100 a 200 kWh

```python
consumo_kwh = 142
tarifa_kwh = 623.5

total_servicios = consumo_kwh * tarifa_kwh

consumo_mayor_igual_100 = consumo_kwh >= 100
consumo_menor_igual_200 = consumo_kwh <= 200

consumo_normal = (
    consumo_mayor_igual_100 and consumo_menor_igual_200
)

print("Total a pagar:", total_servicios)
print("¿Consume al menos 100 kWh?", consumo_mayor_igual_100)
print("¿Consume máximo 200 kWh?", consumo_menor_igual_200)
print("¿El consumo es normal?", consumo_normal)
```

Ejercicio 8
Se calcula el promedio de tres notas y se verifica si el estudiante obtiene una nota de aprobación igual o superior a 3.0

```python
nota1 = 4.2
nota2 = 3.5
nota3 = 4.8

promedio_notas = (nota1 + nota2 + nota3) / 3

aprobado = promedio_notas >= 3.0

print("Promedio de notas:", promedio_notas)
print("¿El estudiante aprobó?", aprobado)
```

Ejercicio 9
Se registra el porcentaje de asistencia y se utiliza el operador not para comprobar que el estudiante no reprobó

```python
asistencia_porcentaje = 85

reprobo = promedio_notas < 3.0

no_reprobo = not (promedio_notas < 3.0)

print("Asistencia:", asistencia_porcentaje, "%")
print("¿El estudiante reprobó?", reprobo)
print("¿El estudiante NO reprobó?", no_reprobo)
```

Ejercicio 10
Se intercambian los valores de los precios del tornillo y la tuerca usando una variable temporal, sin escribir manualmente los nuevos precios

```python
precio_tornillo = 250
precio_tuerca = 180

print("Antes del intercambio:")
print("Precio del tornillo:", precio_tornillo)
print("Precio de la tuerca:", precio_tuerca)

temporal = precio_tornillo
precio_tornillo = precio_tuerca
precio_tuerca = temporal

print("Después del intercambio:")
print("Precio del tornillo:", precio_tornillo)
print("Precio de la tuerca:", precio_tuerca)
```
