Ejercicio 1. Clasificar el stock de un producto

```python
cantidad_en_stock = 3

if cantidad_en_stock == 0:
    nivel = "Agotado"
elif cantidad_en_stock < 5:
    nivel = "Stock bajo"
else:
    nivel = "Stock suficiente"

print(nivel)
```

Ejercicio 2. Alerta de reposición inmediata

```python
if nivel == "Agotado" or cantidad_en_stock < 5:
    print("Reponer pronto")
else:
    print("Sin acción requerida")
```

Ejercicio 5. Aprobación con mensaje

```python
nota_1 = 4.2
nota_2 = 3.5
nota_3 = 4.8

promedio = (nota_1 + nota_2 + nota_3) / 3

if promedio < 3.0:
    resultado = "Reprobado"
elif promedio <= 4.4:
    resultado = "Aprobado"
else:
    resultado = "Aprobado con excelencia"

print("Promedio:", promedio)
print("Resultado:", resultado)
```

Ejercicio 6. Recorrer el inventario completo

```python

cantidades_en_stock = [0, 3, 12, 0, 7, 2]

for indice in range(len(cantidades_en_stock)):
    cantidad = cantidades_en_stock[indice]

    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    print(f"Producto {indice}: {nivel}")
```

Ejercicio 7. Promedio semanal de temperaturas

```python
temperaturas_semana = [18.5, 21.0, 19.8, 24.3, 17.2]

suma_temperaturas = 0
dias_calurosos = 0

for indice in range(len(temperaturas_semana)):
    temperatura = temperaturas_semana[indice]
    suma_temperaturas += temperatura

    if temperatura > 20:
        dias_calurosos += 1

promedio_semanal = suma_temperaturas / len(temperaturas_semana)

print("Promedio semanal:", promedio_semanal)
print("Días calurosos:", dias_calurosos)
```

Ejercicio 8. Comprar tornillos mientras alcance el presupuesto

```python
presupuesto_ferreteria = 3000
precio_tornillo = 250
tornillos_comprados = 0

while presupuesto_ferreteria >= precio_tornillo:
    presupuesto_ferreteria -= precio_tornillo
    tornillos_comprados += 1

print("Tornillos comprados:", tornillos_comprados)
print("Presupuesto restante:", presupuesto_ferreteria)
```

Ejercicio 9. Buscar la primera herramienta cara

```python
precios_herramientas = [12000.0, 8500.0, 25000.0, 9800.0]

for precio in precios_herramientas:
    if precio > 20000:
        print(f"Se encontró la primera herramienta cara: ${precio}")
        break
    else:
        print(f"Se sigue revisando el precio: ${precio}")
```

ercicio 10. Revisión completa del inventario de la ferretería

```python
nombres_herramientas = [
    "Martillo", "Taladro", "Sierra",
    "Destornillador", "Nivel", "Llave inglesa"
]

cantidades_herramientas = [5, 0, 2, 8, 0, 1]

for indice in range(len(nombres_herramientas)):
    nombre = nombres_herramientas[indice]
    cantidad = cantidades_herramientas[indice]

    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    if nivel == "Stock suficiente":
        continue
    elif nivel == "Agotado":
        print(f"Reposición urgente: {nombre}")
        break
    else:
        print(f"Stock bajo: {nombre}, cantidad: {cantidad}")
```
